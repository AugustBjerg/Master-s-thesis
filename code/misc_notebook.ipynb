```python
import numpy as np
import plotly.graph_objects as go

# ---------------------------
# Theoretical speed-power data
# ---------------------------
speed_knots = np.linspace(0, 16, 300)

# Choose coefficient so that:
# at 5 kn -> 500 kW
# at 10 kn -> 4000 kW (8x higher)
k = 4.0
power_kw = k * speed_knots**3

# Example service points
service_speeds = np.array([5, 10])
service_powers = k * service_speeds**3

# ---------------------------
# Figure
# ---------------------------
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=speed_knots,
        y=power_kw,
        mode="lines",
        name="Theoretical curve",
        line=dict(color="#1f77b4", width=4),
        fill="tozeroy",
        fillcolor="rgba(31, 119, 180, 0.12)",
        hovertemplate="Speed: %{x:.1f} kn<br>Power: %{y:,.0f} kW<extra></extra>",
    )
)

fig.add_trace(
    go.Scatter(
        x=service_speeds,
        y=service_powers,
        mode="markers",
        name="Example service points",
        marker=dict(
            size=11,
            color="#d62728",
            line=dict(color="white", width=1.5)
        ),
        hovertemplate="Speed: %{x:.1f} kn<br>Power: %{y:,.0f} kW<extra></extra>",
    )
)

fig.add_annotation(
    x=5,
    y=500,
    text="5 kn, 500 kW",
    showarrow=True,
    arrowhead=2,
    ax=-35,
    ay=-45,
    font=dict(size=12),
)

fig.add_annotation(
    x=10,
    y=4000,
    text="10 kn, 4000 kW",
    showarrow=True,
    arrowhead=2,
    ax=35,
    ay=-45,
    font=dict(size=12),
)

# ---------------------------
# Layout styling
# ---------------------------
fig.update_layout(
    title="Theoretical Speed-Power Curve",
    template="plotly_white",
    font=dict(family="Arial", size=14),
    width=950,
    height=550,
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="center",
        x=0.5
    ),
    margin=dict(l=70, r=40, t=80, b=70),
    paper_bgcolor="white",
    plot_bgcolor="white",
)

fig.update_xaxes(
    title="Speed (knots)",
    range=[0, 16],
    showgrid=True,
    gridcolor="rgba(0,0,0,0.08)",
    zeroline=False,
    ticks="outside",
    showline=True,
    linecolor="black"
)

fig.update_yaxes(
    title="Delivered Power (kW)",
    showgrid=True,
    gridcolor="rgba(0,0,0,0.08)",
    zeroline=False,
    ticks="outside",
    showline=True,
    linecolor="black"
)

fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go

# ---------------------------
# Theoretical speed-power data
# ---------------------------
speed_knots = np.linspace(0, 16, 300)

# Clean-hull theoretical curve
k = 4.0
power_clean_kw = k * speed_knots**3

# Illustrative efficiency deterioration / fouling shift
shift_factor = 1.15
power_shifted_kw = shift_factor * power_clean_kw

# ---------------------------
# Figure
# ---------------------------
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=speed_knots,
        y=power_clean_kw,
        mode="lines",
        name="Clean condition",
        line=dict(color="#1f77b4", width=4),
        hovertemplate="Speed: %{x:.1f} kn<br>Power: %{y:,.0f} kW<extra></extra>",
    )
)

fig.add_trace(
    go.Scatter(
        x=speed_knots,
        y=power_shifted_kw,
        mode="lines",
        name="Deteriorated condition",
        line=dict(color="#d62728", width=4, dash="dash"),
        hovertemplate="Speed: %{x:.1f} kn<br>Power: %{y:,.0f} kW<extra></extra>",
    )
)

# Optional shaded band between the curves
fig.add_trace(
    go.Scatter(
        x=np.concatenate([speed_knots, speed_knots[::-1]]),
        y=np.concatenate([power_clean_kw, power_shifted_kw[::-1]]),
        fill="toself",
        fillcolor="rgba(214, 39, 40, 0.10)",
        line=dict(color="rgba(255,255,255,0)"),
        hoverinfo="skip",
        showlegend=False,
    )
)

# Annotation showing the shift
x_anno = 11
y_clean_anno = k * x_anno**3
y_shift_anno = shift_factor * y_clean_anno

fig.add_annotation(
    x=x_anno,
    y=y_shift_anno,
    text="Efficiency loss / curve shift",
    showarrow=True,
    arrowhead=2,
    ax=60,
    ay=-40,
    font=dict(size=12)
)

# ---------------------------
# Layout styling
# ---------------------------
fig.update_layout(
    title="Theoretical Speed-Power Curve",
    template="plotly_white",
    font=dict(family="Arial", size=14),
    width=950,
    height=550,
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="center",
        x=0.5
    ),
    margin=dict(l=70, r=40, t=80, b=70),
    paper_bgcolor="white",
    plot_bgcolor="white",
)

fig.update_xaxes(
    title="Speed (knots)",
    range=[0, 16],
    showgrid=True,
    gridcolor="rgba(0,0,0,0.08)",
    zeroline=False,
    ticks="outside",
    showline=True,
    linecolor="black"
)

fig.update_yaxes(
    title="Delivered Power (kW)",
    showgrid=True,
    gridcolor="rgba(0,0,0,0.08)",
    zeroline=False,
    ticks="outside",
    showline=True,
    linecolor="black"
)

fig.show()
```
